```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python

wanted_columns = ['Player', 'Pos', 'Squad', 'Comp', 'Gls', 'Ast', 'G+A', 'Saves', 'Tkl+Int', 'KP']

df = pd.read_csv('players_data_light-2024_2025.csv', usecols=wanted_columns)

df
```

```python
# Players  per League (only top 5 European Leagues)

print('\nPlayers per League: ')
print(df['Comp'].value_counts())

# Players distributed by Position

df['Pos'] = df['Pos'].apply(lambda pos : pos.split(',')[0])

print('\nDistribution by Position: ')
print(df['Pos'].value_counts())
```

```python
# Top Scorers (Top 10 players with most Goals)

top_scorers = df.sort_values('Gls', ascending=False).head(10)[['Player', 'Squad', 'Comp', 'Pos', 'Gls']]
print(top_scorers)

plt.figure(figsize=(10, 5))
sns.barplot(data=top_scorers, x='Gls', y='Player', hue='Comp')
plt.show()
```

```python
# Top Assistants (Top 10 players with most Assists)

top_assistants = df.sort_values('Ast', ascending=False).head(10)[['Player', 'Squad', 'Comp', 'Pos', 'Ast']]
print(top_assistants)

plt.figure(figsize=(10,5))
sns.barplot(data=top_assistants, x='Ast', y= 'Player', hue='Comp')
plt.show()
```

```python
# Top offensive players (Top 10 players with most G+A, Goals + Assists)

top_offensiveNumbers = df.sort_values('G+A', ascending=False).head(10)[['Player', 'Squad', 'Comp', 'Pos', 'G+A']]
print(top_offensiveNumbers)

plt.figure(figsize=(10,5))
sns.barplot(data=top_offensiveNumbers, x='G+A', y='Player', hue='Comp')
plt.show()
```

```python
# Top defensive players (Top 10 players with most Tackles + Interceptions)

top_defensiveNumbers = df.sort_values('Tkl+Int', ascending=False).head(10)[['Player', 'Squad', 'Comp', 'Pos', 'Tkl+Int']]
print(top_defensiveNumbers)

plt.figure(figsize=(10,5))
sns.barplot(data=top_defensiveNumbers, x='Tkl+Int', y='Player', hue='Comp')
plt.show()
```

```python
# Top Goalkeepers with more Saves (Top 10 Goalkeepers with more Saves)

top_Goalkeepers = df.sort_values('Saves', ascending=False).head(10)[['Player', 'Squad', 'Comp', 'Pos', 'Saves']]
print(top_Goalkeepers)

plt.figure(figsize=(10, 5))
sns.barplot(data=top_Goalkeepers, x='Saves', y='Player', hue='Comp')
plt.show()
```

```python
# League with better defensive statistics

df.groupby('Comp')['Tkl+Int'].mean().round(2).sort_values(ascending=False)
```

```python
# League with better offensive statistics

df.groupby('Comp')['G+A'].mean().round(2).sort_values(ascending=False)
```

```python
# grouped by position

goalkeepers = df[df['Pos'] == 'GK']
defenders = df[df['Pos'] == 'DF']
midfielders = df[df['Pos'] == 'MF']
fowards = df[df['Pos'] == 'FW']
```

```python
# best players by position, formation = 3-4-3 (1 Goalkeeper, 3 Defenders, 4 Midfielders, 3 Fowards)

best_goalkeeper = goalkeepers.sort_values('Saves', ascending=False).head(1)[['Player', 'Pos', 'Squad', 'Comp']]
display(best_goalkeeper)

best_defenders = defenders.sort_values('Tkl+Int', ascending=False).head(3)[['Player', 'Pos', 'Squad', 'Comp']]
display(best_defenders)

best_midfielders = midfielders.sort_values('KP', ascending=False).head(4)[['Player', 'Pos', 'Squad', 'Comp']]
display(best_midfielders)

best_fowards = fowards.sort_values('G+A', ascending=False).head(3)[['Player', 'Pos', 'Squad', 'Comp']]
display(best_fowards)
```

```python
# Dream Team, the best players by each statistic (Gk = Saves, Df = Tkl+Int, Mf = Key Passes, Fw = G+A)

dream_team = pd.concat([best_goalkeeper, best_defenders, best_midfielders, best_fowards])

print('\n DREAM TEAM SEASON 2024/25')
display(dream_team)
```
